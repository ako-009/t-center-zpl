```python
"""
T-CENTER IN SILICON — COMPLETE IMPLEMENTATION (CORRECTED)
Phonons -> Huang-Rhys Factor -> Luminescence Spectrum

This is the FIXED version. Three real bugs have been corrected
(each marked "FIX #n" below), and one known limitation is clearly
disclosed (not silently hidden):

  FIX #1: Earlier bond-distance checks used plain (non-periodic)
          distance, which wrongly suggested Carbon was disconnected
          from the Si lattice. Using PROPER periodic (PBC-aware)
          distance confirms the geometry is actually correct
          (C-Si ~1.97 A, C-H ~1.15 A) -- see the bond check printed
          in Step 2.

  FIX #2: The displacement step size for finite-difference forces
          was small enough (0.01 A) to pick up numerical noise from
          MACE near the substituted Carbon atom, producing
          unphysical high-frequency "ghost" modes (115/293 meV --
          no real Si/C/H phonon goes that high). Increased to
          0.03 A, and added an automatic filter that excludes any
          mode above 80 meV from the Huang-Rhys sum as a safety net.

  FIX #3: The "experimental" calibration target was S_expt = 0.55,
          which does NOT match the literature. The real published
          experimental Huang-Rhys factor for the T-center is
          S = 1.47 (Bergeron et al., PRX Quantum 2020 -- the actual
          reference cited by the ACS Nano MACE-phonon paper). This
          is now corrected, with the paper's own DFT-calculated
          value (S=1.88) also shown for honest comparison.

  KNOWN LIMITATION (disclosed, not fixed): the paper uses the
  mace-omat-0-medium foundation model (Open Materials 2024
  dataset). The installed mace-torch version here only exposes
  mace-mp-0 ("small"/"medium"/"large", Materials Project dataset)
  through mace_mp(). This is a different but comparable foundation
  model; results are qualitatively similar but won't numerically
  match the paper's exact published values. Upgrading mace-torch
  to access mace-omat-0 is a clear, well-defined next step.

Also note: Phonopy's primitive_matrix is intentionally left at its
default here. An earlier attempt to force primitive_matrix='P'
was tested; if your Phonopy version raises an error with that
argument, simply remove it -- the supercell_matrix=identity setting
already ensures the full defect cell is used for the dynamical
matrix, which is what matters physically.

A built-in safety check (Step 3.5) automatically flags any leftover
unphysical high-frequency phonon modes, so you see immediately if
something looks wrong, instead of finding out later from a strange
final number.
"""

import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'TRUE'

import torch
from mace import modules
torch.serialization.add_safe_globals([modules.models.ScaleShiftMACE])

import numpy as np
import matplotlib.pyplot as plt
from ase.build import bulk, make_supercell
from ase.optimize import BFGS
from ase import Atom, Atoms
from ase.geometry import get_distances
from mace.calculators import mace_mp
from phonopy import Phonopy
from phonopy.structure.atoms import PhonopyAtoms

print("All imports successful")
print("=" * 60)

# IMPORTANT: This installed version of mace-torch does NOT expose
# 'mace-omat-0-medium' through the mace_mp() function -- checking
# inspect.signature(mace_mp) confirms only "small", "medium", or
# "large" are valid model strings here, and the docstring states
# this loads the Materials Project (MPtrj) trained model, not the
# Open Materials 2024 model used in the ACS Nano paper.
#
# This is a KNOWN, DISCLOSED limitation, not a bug: the paper uses
# mace-omat-0-medium (PBE+U, Open Materials 2024 dataset); this
# notebook uses mace-mp-0 medium (PBE, Materials Project dataset).
# Both are reasonable general-purpose foundation models; results
# should be qualitatively similar but will not numerically match
# the paper exactly. Upgrading mace-torch to a version that bundles
# mace-omat-0 is the correct Week 2 fix.
MACE_MODEL = 'medium'


# ============================================================
# STEP 1: BUILD T-CENTER STRUCTURE
# T-center = Carbon substitutional + Hydrogen interstitial in Si
# ============================================================

print("\nSTEP 1: Building T-center structure...")

si = bulk('Si', 'diamond', a=5.43)
P = np.diag([3, 3, 3])
si_super = make_supercell(si, P)
print(f"  Pristine supercell: {len(si_super)} Si atoms")

# Replace atom 0 with Carbon (this IS the lattice site -- substitutional)
si_super[0].symbol = 'C'
c_position = si_super[0].position

# Find nearest Si neighbor using PERIODIC distance (FIX #1: a plain,
# non-periodic distance calculation can wrongly suggest an atom is
# far away when it is actually close through the periodic boundary)
positions = si_super.get_positions()
cell_vectors = si_super.get_cell()
_, pbc_dists = get_distances(
    [c_position], positions, cell=cell_vectors, pbc=True)
pbc_dists = pbc_dists[0]
pbc_dists[0] = np.inf
nearest_idx = np.argmin(pbc_dists)

direction = positions[nearest_idx] - c_position
# wrap direction into the correct periodic image
direction = direction - cell_vectors.T @ np.round(
    np.linalg.solve(cell_vectors.T, direction))
direction = direction / np.linalg.norm(direction)

CH_BOND_LENGTH = 1.10
h_position = c_position + direction * CH_BOND_LENGTH
si_super.append(Atom('H', position=h_position))

symbols = si_super.get_chemical_symbols()
print(f"  Total atoms: {len(si_super)}  "
      f"(Si={symbols.count('Si')}, C={symbols.count('C')}, "
      f"H={symbols.count('H')})")

t_center = si_super.copy()
print("  T-center structure built")


# ============================================================
# STEP 2: RELAX WITH MACE
# ============================================================

print("\nSTEP 2: Relaxing T-center with MACE...")

t_center.calc = mace_mp(model=MACE_MODEL, device='cpu',
                         default_dtype='float64')

forces_before = t_center.get_forces()
print(f"  Max force BEFORE relaxation: "
      f"{np.abs(forces_before).max():.4f} eV/A")

opt = BFGS(t_center, logfile=None)
opt.run(fmax=0.001)   # tightened from 0.01 -> 0.001

forces_after = t_center.get_forces()
print(f"  Max force AFTER  relaxation: "
      f"{np.abs(forces_after).max():.6f} eV/A")
print(f"  Structure relaxed")

t_relaxed = t_center.copy()
t_relaxed.calc = None

# Verify bonding using PBC-aware distances (sanity check before
# trusting anything downstream)
positions_final = t_relaxed.get_positions()
cell_final = t_relaxed.get_cell()
c_pos_final = positions_final[0:1]
_, check_dists = get_distances(
    c_pos_final, positions_final, cell=cell_final, pbc=True)
check_dists = check_dists[0]
check_dists[0] = np.inf
symbols_final = t_relaxed.get_chemical_symbols()

order = np.argsort(check_dists)
ch_dist = None
nearest_si_dist = None
for idx in order:
    if symbols_final[idx] == 'H' and ch_dist is None:
        ch_dist = check_dists[idx]
    if symbols_final[idx] == 'Si' and nearest_si_dist is None:
        nearest_si_dist = check_dists[idx]
    if ch_dist is not None and nearest_si_dist is not None:
        break

print(f"\n  Bond check (periodic distances):")
print(f"    C-H distance:        {ch_dist:.3f} A  (expect ~1.0-1.15 A)")
print(f"    Nearest C-Si distance: {nearest_si_dist:.3f} A  "
      f"(expect ~1.9-2.0 A)")
if nearest_si_dist > 2.2:
    print(f"    WARNING: Carbon may not be properly bonded to Si!")
else:
    print(f"    Geometry looks physically reasonable.")

# ============================================================
# STEP 2.5: EXPORT FOR EXCITED-STATE DFT (Quantum Espresso)
# ============================================================
# This step saves your relaxed structure in multiple formats and
# writes Quantum Espresso input files for the NEXT phase of the
# project: computing the real excited-state atomic displacement
# (Delta R) via DFT, instead of the calibrated value used below
# in Step 4. This is the concrete plan for removing the
# "disclosed limitation" noted throughout this script.
#
# IMPORTANT: the .pwi files below are READY TEMPLATES, not files
# meant to be executed today. Running them requires: (a) real
# pseudopotential files downloaded for Si, C, H (see comment
# below), (b) a working Quantum Espresso installation, and
# (c) realistically hours-to-days of compute time, especially
# once the HSE06 hybrid functional is used (as in the paper).

print("\nSTEP 2.5: Exporting structure for DFT excited-state calc...")

from ase.io import write

# -- Universal format exports (safe, instant, always useful) --
write('t_center_ground.xyz', t_relaxed)
write('t_center_ground.cif', t_relaxed)

try:
    write('POSCAR_ground', t_relaxed, format='vasp')
    print("  Saved: t_center_ground.xyz, .cif, POSCAR_ground")
except Exception as e:
    print(f"  Saved: t_center_ground.xyz, .cif (POSCAR skipped: {e})")

# -- Quantum Espresso input file templates --
# NOTE: pseudopotential filenames below are PLACEHOLDERS. Before
# actually running QE, download real pseudopotentials for
# Si, C, H from the SSSP efficiency library (PBE functional):
#   https://www.materialscloud.org/discover/sssp/table/efficiency
# and place the .UPF files in a local 'pseudo' folder, updating
# 'pseudo_dir' below to match.

pseudopotentials = {
    'Si': 'Si.pbe-n-kjpaw_psl.1.0.0.UPF',
    'C':  'C.pbe-n-kjpaw_psl.1.0.0.UPF',
    'H':  'H.pbe-kjpaw_psl.1.0.0.UPF',
}

try:
    write('t_center_ground.pwi', t_relaxed, format='espresso-in',
          pseudopotentials=pseudopotentials,
          kpts=(1, 1, 1),     # Gamma-point only, large supercell
          input_data={
              'control': {
                  'calculation': 'relax',
                  'prefix': 't_center_gs',
                  'outdir': './qe_out',
                  'pseudo_dir': './pseudo',
              },
              'system': {
                  'ecutwfc': 40,    # plane-wave cutoff (Ry), typical
                                    # starting value -- needs proper
                                    # convergence testing before
                                    # production use
                  'ecutrho': 160,
                  'occupations': 'fixed',   # NORMAL ground state
              },
              'electrons': {
                  'conv_thr': 1.0e-8,
              },
          })
    print("  Saved: t_center_ground.pwi  (QE ground-state input)")
except Exception as e:
    print(f"  QE ground-state input skipped (espresso writer "
          f"unavailable: {e})")

try:
    write('t_center_excited.pwi', t_relaxed, format='espresso-in',
          pseudopotentials=pseudopotentials,
          kpts=(1, 1, 1),
          input_data={
              'control': {
                  'calculation': 'relax',
                  'prefix': 't_center_es',
                  'outdir': './qe_out',
                  'pseudo_dir': './pseudo',
              },
              'system': {
                  'ecutwfc': 40,
                  'ecutrho': 160,
                  'occupations': 'from_input',   # constrained
                                                    # occupation
                                                    # (delta-SCF)
                  'nspin': 2,    # spin-polarized, required for
                                  # delta-SCF excited states
              },
              'electrons': {
                  'conv_thr': 1.0e-8,
              },
          })
    print("  Saved: t_center_excited.pwi  (QE excited-state, "
          "delta-SCF input)")
    print("  REMINDER: the constrained-occupation band index must")
    print("  be set manually AFTER inspecting ground-state DOS/bands")
    print("  to identify the defect level in the gap.")
except Exception as e:
    print(f"  QE excited-state input skipped: {e}")

print("\n  Next steps (NOT run today -- see guide):")
print("  1. Download real pseudopotentials for Si, C, H (SSSP, PBE)")
print("  2. Run: pw.x < t_center_ground.pwi   (ground-state SCF)")
print("  3. Inspect output to find defect level in the band gap")
print("  4. Edit t_center_excited.pwi with correct band index")
print("  5. Run: pw.x < t_center_excited.pwi  (excited-state relax)")
print("  6. Compute Delta R = positions_excited - positions_ground")
print("  7. Replace the calibrated dR in Step 4 below with this")
print("     real, DFT-derived value")



# ============================================================
# STEP 3: COMPUTE PHONONS WITH PHONOPY + MACE
# ============================================================

print("\nSTEP 3: Computing phonons...")

N = len(t_relaxed)
print(f"  System size: {N} atoms -> {3*N} phonon branches")

cell = PhonopyAtoms(
    symbols=t_relaxed.get_chemical_symbols(),
    positions=t_relaxed.get_positions(),
    cell=t_relaxed.get_cell()
)

try:
    phonon = Phonopy(
        cell,
        supercell_matrix=[[1, 0, 0], [0, 1, 0], [0, 0, 1]],
        primitive_matrix='P'   # forces identity primitive matrix so
                                 # Phonopy treats the full defect
                                 # supercell as-is, instead of
                                 # auto-detecting a smaller "hidden"
                                 # primitive cell that mixes up
                                 # defect modes with bulk modes
    )
    print("  primitive_matrix='P' accepted.")
except Exception as e:
    print(f"  primitive_matrix='P' not accepted by this Phonopy "
          f"version ({e}); falling back to default.")
    phonon = Phonopy(
        cell,
        supercell_matrix=[[1, 0, 0], [0, 1, 0], [0, 0, 1]]
    )

phonon.generate_displacements(distance=0.03)   # FIX #2: 0.01 -> 0.03 A
supercells = phonon.supercells_with_displacements

print(f"  Displaced configs: {len(supercells)}")
print(f"  Computing MACE forces...")

forces = []
for i, sc in enumerate(supercells):
    sc_ase = Atoms(symbols=sc.symbols, positions=sc.positions,
                    cell=sc.cell, pbc=True)
    sc_ase.calc = mace_mp(model=MACE_MODEL, device='cpu',
                           default_dtype='float64')
    f = sc_ase.get_forces()
    forces.append(f)
    if (i + 1) % 20 == 0 or (i + 1) == len(supercells):
        print(f"    {i+1}/{len(supercells)} done")

phonon.forces = forces
phonon.produce_force_constants()
print("  Force constants built")

phonon.run_mesh([10, 10, 10], is_gamma_center=True)
phonon.run_total_dos()
print("  Phonon DOS computed")

print("\nSTEP 3 COMPLETE: Phonon calculation done")


# ============================================================
# STEP 3.5: SANITY CHECK -- catch unphysical modes automatically
# ============================================================

print("\nSTEP 3.5: Checking for unphysical modes...")

phonon.run_qpoints([[0, 0, 0]], with_eigenvectors=True)
qpts = phonon.get_qpoints_dict()
check_freqs_THz = qpts['frequencies'][0]
check_freqs_meV = check_freqs_THz * 4.136

n_imaginary = np.sum(check_freqs_THz < -0.01)
HIGH_FREQ_CEILING_MEV = 80.0
suspicious = check_freqs_meV[check_freqs_meV > HIGH_FREQ_CEILING_MEV]

if n_imaginary > 0:
    print(f"  WARNING: {n_imaginary} imaginary mode(s) -- "
          f"structure may not be fully relaxed.")
else:
    print(f"  No imaginary modes. Good.")

if len(suspicious) > 0:
    print(f"  WARNING: {len(suspicious)} mode(s) above "
          f"{HIGH_FREQ_CEILING_MEV} meV: {np.round(suspicious, 1)} meV")
    print(f"  These will be EXCLUDED from the Huang-Rhys sum below,")
    print(f"  since no real Si/C/H phonon should be this stiff.")
else:
    print(f"  All frequencies within the physically expected range.")


# ============================================================
# STEP 4: HUANG-RHYS FACTOR
# ============================================================

print("\nSTEP 4: Computing Huang-Rhys factors...")

freqs_THz = qpts['frequencies'][0]
raw_evecs = qpts['eigenvectors'][0]
evecs_T = np.real(raw_evecs.T)

# Remove acoustic modes (omega ~ 0, rigid translation)
mask = freqs_THz > 0.1
freqs_THz = freqs_THz[mask]
evecs_T = evecs_T[mask]

# also remove any remaining unphysically high modes from the
# Huang-Rhys sum, so a numerical artifact (if any slips through)
# cannot silently inflate S_total.
freqs_meV_check = freqs_THz * 4.136
good_mask = freqs_meV_check <= HIGH_FREQ_CEILING_MEV
n_removed = np.sum(~good_mask)
if n_removed > 0:
    print(f"  Excluding {n_removed} unphysical mode(s) from S calculation.")
freqs_THz = freqs_THz[good_mask]
evecs_T = evecs_T[good_mask]

n_modes = len(freqs_THz)
print(f"  Modes used in calculation: {n_modes}")
print(f"  Frequency range: {freqs_THz.min():.2f} to "
      f"{freqs_THz.max():.2f} THz  "
      f"({freqs_THz.min()*4.136:.1f} to {freqs_THz.max()*4.136:.1f} meV)")

# Unit conversion: S_i = omega_i * |dq_i|^2 / (2*hbar)
conv = (2 * np.pi * 1e12 * 1.66054e-27 * 1e-20) / (2 * 1.05457e-34)
print(f"\n  Conversion factor: {conv:.5f}")

# FIX #3: correct experimental target value.
# S_expt = 1.47 is the REAL experimental Huang-Rhys factor for the
# T-center, from Bergeron et al., PRX Quantum 1, 020301 (2020) --
# the actual reference cited by the ACS Nano MACE-phonon paper.
# The paper's OWN full-DFT calculation gets S = 1.88 (shown below
# for comparison). The OLD value of 0.55 used previously does not
# correspond to any value in the literature for this defect.
S_expt = 1.47
S_paper_dft = 1.88

masses = t_relaxed.get_masses()
mass_vec = np.repeat(masses, 3)
sqrt_mass = np.sqrt(mass_vec)

print(f"\n  Scanning dR to match S_expt = {S_expt} "
      f"(Bergeron et al. 2020)...")
print(f"  {'dR (A)':>8}  {'S_total':>9}  {'ZPL%':>8}")
print(f"  {'-'*32}")

best_dR = 0.20
best_diff = 999.0
best_S = 0.0

for dR_test in np.arange(0.05, 0.80, 0.005):
    delta_R_test = np.zeros((N, 3))
    delta_R_test[0, 2] = dR_test
    delta_R_flat = delta_R_test.flatten()
    delta_Q = delta_R_flat * sqrt_mass
    dq_proj = evecs_T @ delta_Q
    S_i_test = conv * freqs_THz * dq_proj**2
    S_test = np.sum(S_i_test)
    diff = abs(S_test - S_expt)
    if diff < best_diff:
        best_diff = diff
        best_dR = dR_test
        best_S = S_test
    if abs(dR_test % 0.05) < 0.003:
        print(f"  {dR_test:>8.3f}  {S_test:>9.4f}  "
              f"{np.exp(-S_test)*100:>7.1f}%")

print(f"\n  Best match: dR = {best_dR:.3f} A -> "
      f"S = {best_S:.4f} (target {S_expt})")
print(f"  Physical meaning: C atom moves {best_dR:.3f} A "
      f"when electron excites")
print(f"  NOTE: dR is CALIBRATED to match experiment here, not")
print(f"  independently derived from excited-state DFT. The real")
print(f"  Paper 1 method computes dR directly from HSE06 ground-")
print(f"  and excited-state relaxations (their eq. 1).")

# Final computation with calibrated dR
delta_R = np.zeros((N, 3))
delta_R[0, 2] = best_dR
delta_R_flat = delta_R.flatten()
delta_Q = delta_R_flat * sqrt_mass

dQ_norm_sq = np.dot(delta_Q, delta_Q)
dq_proj = evecs_T @ delta_Q
parseval_sum = np.sum(dq_proj**2)
completeness = 100 * parseval_sum / dQ_norm_sq
print(f"\n  Parseval completeness: {completeness:.1f}%")

S_i = conv * freqs_THz * dq_proj**2
S_total = np.sum(S_i)
ZPL_frac = np.exp(-S_total)

print(f"\n  S_total            = {S_total:.4f}")
print(f"  ZPL frac           = {ZPL_frac*100:.1f}%")
print(f"  Experimental target: S={S_expt} (Bergeron et al. 2020)")
print(f"  Paper's own DFT calc: S={S_paper_dft}, "
      f"ZPL={np.exp(-S_paper_dft)*100:.0f}%")

top_idx = np.argsort(S_i)[::-1][:10]
print(f"\n  Top 10 modes by Huang-Rhys factor:")
print(f"  {'Freq(THz)':>10}  {'Freq(meV)':>10}  "
      f"{'S_i':>8}  {'% of S_tot':>10}")
for idx in top_idx:
    print(f"  {freqs_THz[idx]:>10.2f}  {freqs_THz[idx]*4.136:>10.1f}  "
          f"{S_i[idx]:>8.5f}  {S_i[idx]/S_total*100:>9.1f}%")

print("\nSTEP 4 COMPLETE: Huang-Rhys factors computed")


# ============================================================
# STEP 5: SPECTRAL DENSITY S(hw)
# ============================================================

print("\nSTEP 5: Building spectral density S(hw)...")

freq_eV = freqs_THz * 4.136e-3
sigma = 0.003

omega_ax = np.linspace(0, freq_eV.max() * 1.15, 1000)
S_omega = np.zeros_like(omega_ax)
for s_i, w_i in zip(S_i, freq_eV):
    if s_i > 1e-9:
        S_omega += s_i * np.exp(-(omega_ax - w_i)**2 / (2 * sigma**2))

print(f"  Spectral density built")
print(f"  Peak S(hw) at: {omega_ax[np.argmax(S_omega)]*1000:.1f} meV")


# ============================================================
# STEP 6: LUMINESCENCE SPECTRUM L(E)
# ============================================================

print("\nSTEP 6: Building luminescence spectrum...")

E_ZPL = 0.935   # eV, T-center ZPL (~1326 nm), from literature
E_axis = np.linspace(E_ZPL - 0.35, E_ZPL + 0.03, 5000)
L = np.zeros_like(E_axis)
sigma_ZPL = sigma / 3

L += ZPL_frac * np.exp(-(E_axis - E_ZPL)**2 / (2 * sigma_ZPL**2))

for s_i, w_i in zip(S_i, freq_eV):
    if s_i < 1e-9:
        continue
    L += s_i * ZPL_frac * np.exp(
        -(E_axis - (E_ZPL - w_i))**2 / (2 * sigma**2))

for s1, w1 in zip(S_i, freq_eV):
    if s1 < 1e-3:
        continue
    for s2, w2 in zip(S_i, freq_eV):
        if s2 < 1e-3:
            continue
        weight = (s1 * s2 / 2.0) * ZPL_frac
        if weight < 1e-7:
            continue
        L += weight * np.exp(
            -(E_axis - (E_ZPL - w1 - w2))**2 / (2 * sigma**2))

if L.max() > 0:
    L /= L.max()

print(f"  Luminescence spectrum built")
print(f"  ZPL contribution: {ZPL_frac*100:.1f}% of emission")


# ============================================================
# STEP 7: PLOTS
# ============================================================

print("\nSTEP 7: Generating plots...")

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

dos_dict = phonon.get_total_dos_dict()
axes[0].plot(dos_dict['frequency_points'], dos_dict['total_dos'],
             'b-', linewidth=2)
axes[0].set_xlabel('Frequency (THz)', fontsize=12)
axes[0].set_ylabel('DOS (arb. units)', fontsize=12)
axes[0].set_title('T-Center Phonon DOS\n(Si bulk + C/H local modes)',
                    fontsize=11)
axes[0].set_xlim(left=0)
axes[0].grid(True, alpha=0.3)
local_modes = freqs_THz[S_i > S_i.max() * 0.3]
for f in local_modes:
    axes[0].axvline(x=f, color='red', alpha=0.5,
                     linestyle='--', linewidth=1)

axes[1].fill_between(omega_ax * 1000, S_omega, alpha=0.6,
                       color='steelblue')
axes[1].set_xlabel('hw (meV)', fontsize=12)
axes[1].set_ylabel('S(hw) (arb. units)', fontsize=12)
axes[1].set_title('Spectral Density S(hw)', fontsize=11)
axes[1].grid(True, alpha=0.3)
axes[1].text(0.05, 0.88,
             f'S_total = {S_total:.3f}\nZPL = {ZPL_frac*100:.1f}%\n'
             f'Expt: S={S_expt}',
             transform=axes[1].transAxes, fontsize=10,
             bbox=dict(boxstyle='round', facecolor='wheat'))

axes[2].plot(E_axis - E_ZPL, L, 'r-', linewidth=2)
axes[2].axvline(0, color='gray', linestyle='--', alpha=0.8,
                  linewidth=1.5, label=f'ZPL ({ZPL_frac*100:.1f}%)')
axes[2].fill_between(E_axis - E_ZPL, L, alpha=0.2, color='red')
axes[2].set_xlabel('E - E_ZPL (eV)', fontsize=12)
axes[2].set_ylabel('Luminescence (arb. units)', fontsize=12)
axes[2].set_title('T-Center Luminescence Spectrum', fontsize=11)
axes[2].legend(fontsize=11)
axes[2].set_xlim(-0.35, 0.03)
axes[2].grid(True, alpha=0.3)

plt.suptitle(
    'T-Center in Silicon: Phonons -> Huang-Rhys -> Luminescence (Fixed)\n'
    f'S_total = {S_total:.4f}, ZPL = {ZPL_frac*100:.1f}% '
    f'(Experiment: S={S_expt})',
    fontsize=13, fontweight='bold')
plt.tight_layout()
plt.savefig('t_center_complete_FIXED.png', dpi=150, bbox_inches='tight')
plt.show()


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINAL RESULTS (CORRECTED PIPELINE)")
print("=" * 60)
print(f"System:              T-center in Si ({N} atoms)")
print(f"Phonon modes used:   {n_modes}")
print(f"Freq range:          {freqs_THz.min():.1f} - "
      f"{freqs_THz.max():.1f} THz")
print(f"")
print(f"Calibrated dR:       {best_dR:.3f} A  (C atom, z-dir)")
print(f"Basis completeness:  {completeness:.1f}%")
print(f"")
print(f"S_total (calculated): {S_total:.4f}")
print(f"S (experiment):       {S_expt}  [Bergeron et al. 2020]")
print(f"S (paper's full DFT): {S_paper_dft}  [ACS Nano 2026]")
print(f"")
print(f"ZPL fraction (calc):  {ZPL_frac*100:.1f}%")
print(f"ZPL (experiment):     {np.exp(-S_expt)*100:.0f}%")
print(f"")
print(f"Pipeline:  PHONONS -> HUANG-RHYS -> LUMINESCENCE")
print(f"Method:    MACE foundation model (replaces DFT)")
print(f"Known limitation: dR is calibrated, not independently")
print(f"  derived from excited-state DFT (this is the planned")
print(f"  Week 2 improvement).")
print("=" * 60)
```

```python
from ase.io import read, write
from ase.geometry import get_distances
import numpy as np

# Read from SAME source format
# Both must come from QE to have same coordinate convention

ground_atoms  = read('scf.out',
                     format='espresso-out',
                     index=-1)          # ground from QE output
excited_atoms = read('excited_v2.out',
                     format='espresso-out',
                     index=-1)          # excited from QE output

print(f"Ground  atoms: {len(ground_atoms)}")
print(f"Excited atoms: {len(excited_atoms)}")
print(f"\nGround  cell:\n{ground_atoms.get_cell()}")
print(f"\nExcited cell:\n{excited_atoms.get_cell()}")

# Check first few atom positions
print("\nFirst 3 atom positions comparison:")
print(f"{'Atom':<6} {'Ground':>20} {'Excited':>20}")
gpos = ground_atoms.get_positions()
epos = excited_atoms.get_positions()
syms = ground_atoms.get_chemical_symbols()
for i in range(3):
    print(f"{syms[i]:<6} "
          f"{str(gpos[i].round(4)):>20} "
          f"{str(epos[i].round(4)):>20}")

# Compute ΔR using PBC-aware distances
# This handles coordinate frame differences
cell = ground_atoms.get_cell()
delta_R_real = np.zeros((len(ground_atoms), 3))

for i in range(len(ground_atoms)):
    diff = epos[i] - gpos[i]
    # Wrap into nearest image (handles PBC frame shifts)
    diff = diff - cell.T @ np.round(
        np.linalg.solve(cell.T, diff))
    delta_R_real[i] = diff

# Report
magnitudes = np.linalg.norm(delta_R_real, axis=1)

print(f"\nTop 5 displaced atoms (PBC-corrected):")
print(f"{'Idx':>5}  {'Atom':>5}  {'|ΔR| (Å)':>10}")
print("-"*25)
for idx in np.argsort(magnitudes)[::-1][:5]:
    print(f"  {idx:3d}    {syms[idx]:2s}    {magnitudes[idx]:.4f} Å")

print(f"\nC atom (index 0): |ΔR| = {magnitudes[0]:.4f} Å")
print(f"H atom (last)   : |ΔR| = {magnitudes[-1]:.4f} Å")

if magnitudes[0] < 1.0:
    print(f"\n✓ PHYSICAL — C moved {magnitudes[0]:.3f} Å")
    print(f"  Ready to compute S_total")
    np.save('delta_R_real_corrected.npy', delta_R_real)
    print(f"  Saved: delta_R_real_corrected.npy")
else:
    print(f"\n✗ Still too large ({magnitudes[0]:.3f} Å)")
    print(f"  Checking atom ordering...")
    # Check if atom ordering differs between files
    g_syms = ground_atoms.get_chemical_symbols()
    e_syms = excited_atoms.get_chemical_symbols()
    mismatches = sum(1 for a,b in zip(g_syms, e_syms) if a!=b)
    print(f"  Atom ordering mismatches: {mismatches}")
    if mismatches > 0:
        print(f"  ORDERING MISMATCH — atoms in different order!")
        print(f"  Ground: {g_syms[:5]}")
        print(f"  Excited: {e_syms[:5]}")
```
